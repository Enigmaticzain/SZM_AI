```python

import json
import re
import random
from pathlib import Path

SOURCE = Path("/content/szm-trained-local-corpus.json")
OUT = Path("/content/szm_training_data")
OUT.mkdir(exist_ok=True)

with SOURCE.open(encoding="utf-8") as f:
    corpus = json.load(f)

documents = corpus["documents"]
assert documents, "No documents found in corpus"

# Clean text while preserving meaningful paragraph boundaries.
valid_docs = []
for doc in documents:
    text = doc.get("extractedText", "")
    text = text.replace("\x00", "")
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n{3,}", "\n\n", text).strip()

    if len(text) >= 300:
        valid_docs.append({
            "source": doc.get("fileName", "unknown"),
            "text": text
        })

# Split by source document before chunking to avoid leakage.
random.Random(42).shuffle(valid_docs)
if len(valid_docs) < 5:
    raise ValueError("Too few usable documents for a reliable split")

split_at = max(1, int(len(valid_docs) * 0.9))
split_at = min(split_at, len(valid_docs) - 1)
train_docs = valid_docs[:split_at]
eval_docs = valid_docs[split_at:]

def make_chunks(documents, chunk_size=1800, overlap=200):
    chunks = []
    for doc in documents:
        text = doc["text"]
        start = 0

        while start < len(text):
            end = min(start + chunk_size, len(text))

            # Prefer paragraph or sentence boundaries near the target.
            if end < len(text):
                boundary = max(
                    text.rfind("\n", start + chunk_size // 2, end),
                    text.rfind(". ", start + chunk_size // 2, end)
                )
                if boundary > start:
                    end = boundary + 1

            piece = text[start:end].strip()
            if len(piece) >= 200:
                chunks.append({
                    "source": doc["source"],
                    "text": f"Source: {doc['source']}\n\n{piece}"
                })

            if end >= len(text):
                break
            start = max(start + 1, end - overlap)

    return chunks

train_chunks = make_chunks(train_docs)
eval_chunks = make_chunks(eval_docs)

assert train_chunks and eval_chunks, "Training or evaluation set is empty"

for filename, records in [
    ("train.jsonl", train_chunks),
    ("eval.jsonl", eval_chunks)
]:
    with (OUT / filename).open("w", encoding="utf-8") as f:
        for record in records:
            f.write(json.dumps(record, ensure_ascii=False) + "\n")

print("=== SZM DATASET REPORT ===")
print("Usable source documents:", len(valid_docs))
print("Training source documents:", len(train_docs))
print("Evaluation source documents:", len(eval_docs))
print("Training chunks:", len(train_chunks))
print("Evaluation chunks:", len(eval_chunks))
print("Training characters:", sum(len(x["text"]) for x in train_chunks))
print("Evaluation characters:", sum(len(x["text"]) for x in eval_chunks))
print("Output directory:", OUT)
print("Files:", [p.name for p in OUT.iterdir()])
print("\nDataset preparation complete.")
```

```python

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig

MODEL_NAME = "Qwen/Qwen2.5-1.5B-Instruct"

assert torch.cuda.is_available(), "GPU is unavailable. Select a GPU runtime."

# 4-bit quantization reduces GPU memory requirements.
quant_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True,
)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=quant_config,
    device_map="auto",
)

model.config.use_cache = False

print("Model:", MODEL_NAME)
print("Model loaded successfully")
print("GPU allocated:", round(torch.cuda.memory_allocated() / 1024**3, 2), "GB")
print("Tokenizer vocabulary:", len(tokenizer))
```

```python

import json
import shutil
from pathlib import Path

from datasets import Dataset
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training
from transformers import (
    Trainer,
    TrainingArguments,
    DataCollatorForLanguageModeling,
)

PROJECT = Path("/content/drive/MyDrive/SZM_AI_Training")
TRAIN_FILE = PROJECT / "train.jsonl"
EVAL_FILE = PROJECT / "eval.jsonl"
OUTPUT = Path("/content/szm-ai-lora")

assert TRAIN_FILE.exists() and EVAL_FILE.exists(), "Dataset files missing"

def load_records(path):
    records = []
    with path.open(encoding="utf-8") as f:
        for line in f:
            row = json.loads(line)
            text = row.get("text", "").strip()
            if text:
                records.append({"text": text + tokenizer.eos_token})
    return records

train_records = load_records(TRAIN_FILE)
eval_records = load_records(EVAL_FILE)

assert train_records and eval_records, "Dataset is empty"

# Keep training sequences short for the first T4 experiment.
MAX_LENGTH = 512

def tokenize_batch(batch):
    return tokenizer(
        batch["text"],
        truncation=True,
        max_length=MAX_LENGTH,
        padding=False,
    )

train_ds = Dataset.from_list(train_records).map(
    tokenize_batch, batched=True, remove_columns=["text"]
)
eval_ds = Dataset.from_list(eval_records).map(
    tokenize_batch, batched=True, remove_columns=["text"]
)

model = prepare_model_for_kbit_training(model)
model.gradient_checkpointing_enable()

lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
    target_modules=[
        "q_proj", "k_proj", "v_proj", "o_proj",
        "gate_proj", "up_proj", "down_proj",
    ],
)

model = get_peft_model(model, lora_config)
model.print_trainable_parameters()

args = TrainingArguments(
    output_dir=str(OUTPUT),
    num_train_epochs=1,
    per_device_train_batch_size=1,
    per_device_eval_batch_size=1,
    gradient_accumulation_steps=8,
    learning_rate=2e-4,
    fp16=True,
    logging_steps=10,
    eval_strategy="epoch",
    save_strategy="epoch",
    save_total_limit=1,
    gradient_checkpointing=True,
    report_to="none",
    seed=42,
)

trainer = Trainer(
    model=model,
    args=args,
    train_dataset=train_ds,
    eval_dataset=eval_ds,
    data_collator=DataCollatorForLanguageModeling(
        tokenizer=tokenizer, mlm=False
    ),
    processing_class=tokenizer,
)

print("Training examples:", len(train_ds))
print("Evaluation examples:", len(eval_ds))
print("Maximum sequence length:", MAX_LENGTH)

trainer.train()

# Save the adapter separately; the original base model remains unchanged.
adapter_dir = OUTPUT / "final_adapter"
trainer.save_model(str(adapter_dir))
tokenizer.save_pretrained(str(adapter_dir))

drive_adapter = PROJECT / "szm-ai-lora-adapter"
if drive_adapter.exists():
    shutil.rmtree(drive_adapter)
shutil.copytree(adapter_dir, drive_adapter)

print("Training finished.")
print("Adapter saved to:", drive_adapter)
print("Final metrics:", trainer.evaluate())
```

```python

import gc
import torch
from pathlib import Path
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig
from peft import PeftModel

BASE = "Qwen/Qwen2.5-1.5B-Instruct"
ADAPTER = Path("/content/drive/MyDrive/SZM_AI_Training/szm-ai-lora-adapter")

assert ADAPTER.exists(), f"Adapter not found: {ADAPTER}"

questions = [
    "Explain the purpose of a centrifugal pump.",
    "How does a heat exchanger transfer thermal energy?",
    "What is preventive maintenance, and why is it useful?",
    "What are the engineering trade-offs of carbon fiber composites?",
    "What safety precautions are important when handling lithium-polymer batteries?"
]

quant_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True,
)

def load_base():
    tok = AutoTokenizer.from_pretrained(BASE)
    if tok.pad_token is None:
        tok.pad_token = tok.eos_token

    base = AutoModelForCausalLM.from_pretrained(
        BASE,
        quantization_config=quant_config,
        device_map="auto",
    )
    base.eval()
    return tok, base

def answer(tok, model, question):
    messages = [
        {
            "role": "system",
            "content": (
                "You are SZM AI, an engineering assistant. "
                "Answer clearly. If uncertain, say so."
            ),
        },
        {"role": "user", "content": question},
    ]

    prompt = tok.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )
    inputs = tok(prompt, return_tensors="pt").to("cuda")

    with torch.inference_mode():
        output = model.generate(
            **inputs,
            max_new_tokens=120,
            do_sample=False,
            pad_token_id=tok.eos_token_id,
        )

    new_tokens = output[0, inputs["input_ids"].shape[1]:]
    return tok.decode(new_tokens, skip_special_tokens=True).strip()

# First: evaluate the untouched base model.
tok, base_model = load_base()
base_answers = []

for i, question in enumerate(questions, 1):
    result = answer(tok, base_model, question)
    base_answers.append(result)
    print(f"\n[BASE {i}] {question}\n{result}")

del base_model
gc.collect()
torch.cuda.empty_cache()

# Second: evaluate the same questions using the trained adapter.
tok = AutoTokenizer.from_pretrained(ADAPTER)
if tok.pad_token is None:
    tok.pad_token = tok.eos_token

base_model = AutoModelForCausalLM.from_pretrained(
    BASE,
    quantization_config=quant_config,
    device_map="auto",
)
adapted_model = PeftModel.from_pretrained(base_model, str(ADAPTER))
adapted_model.eval()

adapted_answers = []
for i, question in enumerate(questions, 1):
    result = answer(tok, adapted_model, question)
    adapted_answers.append(result)
    print(f"\n[SZM ADAPTER {i}] {question}\n{result}")

print("\nEvaluation completed. Compare both sets for factual accuracy,")
print("relevance, clarity, and unsupported claims.")
```

```python
import json, hashlib, re
from pathlib import Path

root = Path("/content/drive/MyDrive/SZM_AI_Training")
data = json.loads(
    (root / "szm-trained-local-corpus.json").read_text(encoding="utf-8")
)

seen = {}
manifest = []

for i, doc in enumerate(data["documents"], 1):
    text = doc.get("extractedText", "").strip()
    if not text:
        continue

    digest = hashlib.sha256(text.encode("utf-8")).hexdigest()
    duplicate_of = seen.get(digest)
    if duplicate_of is None:
        seen[digest] = i

    # First readable lines help identify the document's subject.
    lines = [re.sub(r"\s+", " ", x).strip()
             for x in text.splitlines() if x.strip()]
    preview = " ".join(lines[:4])[:350]

    manifest.append({
        "document_id": i,
        "filename": doc.get("fileName") or "NO_FILENAME",
        "characters": len(text),
        "sha256": digest,
        "duplicate_of_document_id": duplicate_of,
        "preview": preview,
    })

out = root / "document_audit_manifest.json"
out.write_text(json.dumps(manifest, indent=2, ensure_ascii=False),
               encoding="utf-8")

print("Documents inspected:", len(manifest))
print("Unique exact-content documents:", len(seen))
print("Exact duplicate copies:", sum(
    x["duplicate_of_document_id"] is not None for x in manifest
))
print("Manifest saved to:", out)

print("\n===== DOCUMENT PREVIEWS =====")
for x in manifest:
    tag = (f" DUPLICATE OF #{x['duplicate_of_document_id']}"
           if x["duplicate_of_document_id"] else "")
    print(f"\n#{x['document_id']} | {x['filename']} | "
          f"{x['characters']:,} chars{tag}")
    print(x["preview"])
```

**Reasoning**:
Perform a programmatically detailed audit and document integrity mapping of the master corpus to identify duplicate content and map out unrelated documents for a clean blacklist.



```python
import json
import hashlib
import re
from pathlib import Path

# 1. Paths
corpus_path = Path("/content/drive/MyDrive/SZM_AI_Training/szm-trained-local-corpus.json")
assert corpus_path.exists(), f"Corpus file not found: {corpus_path}"

# 2. Load master corpus
with corpus_path.open(encoding="utf-8") as f:
    corpus_data = json.load(f)

documents = corpus_data.get("documents", [])
print(f"Loaded {len(documents)} documents from master corpus.\n")

# 3. Analyze duplicate groups using SHA-256
seen_hashes = {}  # hash -> first seen doc_id
duplicates = []
unique_docs = []

for i, doc in enumerate(documents, start=1):
    text = doc.get("extractedText", "").strip()
    doc_id = doc.get("id", f"index_{i}")
    filename = doc.get("fileName") or "NO_FILENAME"

    # Compute SHA-256
    sha256_hash = hashlib.sha256(text.encode("utf-8")).hexdigest()

    if sha256_hash in seen_hashes:
        first_seen_id = seen_hashes[sha256_hash]
        duplicates.append({
            "index": i,
            "id": doc_id,
            "filename": filename,
            "duplicate_of_index": first_seen_id
        })
    else:
        seen_hashes[sha256_hash] = i
        unique_docs.append({
            "index": i,
            "id": doc_id,
            "filename": filename,
            "text": text
        })

print(f"=== DUPLICATE ANALYSIS ===")
print(f"Total duplicates found: {len(duplicates)}")
for dup in duplicates:
    print(f"Document #{dup['index']} ({dup['filename']}) is a duplicate of Document #{dup['duplicate_of_index']}")

# 4. Filter & Blacklist classification
# We will inspect unique documents to classify non-engineering or unrelated content.
# Typical topics to exclude based on prior analysis: general cosmology, general biology (Homo sapiens), English grammar, dictionaries.
non_engineering_keywords = [
    "homo sapiens", "primate", "bipedality",  # Biology
    "universe comprises", "cosmology", "galactic", # Cosmology
    "english grammar", "adverb", "preposition", # Grammar
    "dictionary is a listing", "lexemes"  # Dictionary / General reference
]

blacklist = []
core_engineering = []

print("\n=== CLASSIFICATION & BLACKLIST MAPPING ===")
for u_doc in unique_docs:
    text_lower = u_doc["text"].lower()
    filename_lower = u_doc["filename"].lower()

    is_blacklist = False
    reason = ""

    # Flag general/unrelated texts based on keywords
    for keyword in non_engineering_keywords:
        if keyword in text_lower:
            is_blacklist = True
            reason = f"Matches unrelated domain keyword: '{keyword}'"
            break

    # Flag empty or system diagnostic files (e.g. bash help text or untitled empty documents)
    if len(u_doc["text"].strip()) < 200:
        is_blacklist = True
        reason = "Too short / empty placeholder"
    elif "gnu bash" in text_lower and "shell commands" in text_lower:
        is_blacklist = True
        reason = "Shell environment help output"
    elif "ubuntu desktop guide" in text_lower:
        is_blacklist = True
        reason = "OS general user manual"

    if is_blacklist:
        blacklist.append({
            "index": u_doc["index"],
            "id": u_doc["id"],
            "filename": u_doc["filename"],
            "reason": reason,
            "preview": u_doc["text"][:150].replace('\n', ' ') + "..."
        })
    else:
        core_engineering.append(u_doc)

print(f"Core engineering / relevant documents remaining: {len(core_engineering)}")
print(f"Blacklisted / unrelated documents identified: {len(blacklist)}\n")

print("=== BLACKLISTED DOCUMENTS DETAIL ===")
for item in blacklist:
    print(f"Index: #{item['index']} | File: {item['filename']} | Reason: {item['reason']}")
    print(f"  Preview: {item['preview']}\n")

# Save the audit blacklist mapping
import json
audit_output_path = Path("/content/drive/MyDrive/SZM_AI_Training/document_integrity_blacklist.json")
with audit_output_path.open("w", encoding="utf-8") as f:
    json.dump({
        "blacklist_indices": [item["index"] for item in blacklist],
        "blacklist_details": blacklist,
        "duplicate_details": duplicates
    }, f, indent=2, ensure_ascii=False)

print(f"Integrity audit and blacklist manifest saved to: {audit_output_path}")
```

```python
import json
import hashlib
import re
from pathlib import Path

# 1. Define paths
base_dir = Path("/content/drive/MyDrive/SZM_AI_Training")
base_dir.mkdir(parents=True, exist_ok=True)
corpus_path = base_dir / "szm-trained-local-corpus.json"

# Fallback to local file if Drive path isn't ready
if not corpus_path.exists():
    corpus_path = Path("/content/szm-trained-local-corpus.json")

assert corpus_path.exists(), f"Corpus file not found: {corpus_path}"

# 2. Load master corpus
with corpus_path.open(encoding="utf-8") as f:
    corpus_data = json.load(f)

documents = corpus_data.get("documents", [])

# 3. Analyze duplicate groups using SHA-256 and classify non-engineering content
seen_hashes = {}
duplicates = []
unique_docs = []

for i, doc in enumerate(documents, start=1):
    text = doc.get("extractedText", "").strip()
    doc_id = doc.get("id", f"index_{i}")
    filename = doc.get("fileName") or "NO_FILENAME"
    sha256_hash = hashlib.sha256(text.encode("utf-8")).hexdigest()

    if sha256_hash in seen_hashes:
        first_seen_id = seen_hashes[sha256_hash]
        duplicates.append({
            "index": i,
            "id": doc_id,
            "filename": filename,
            "duplicate_of_index": first_seen_id
        })
    else:
        seen_hashes[sha256_hash] = i
        unique_docs.append({
            "index": i,
            "id": doc_id,
            "filename": filename,
            "text": text
        })

non_engineering_keywords = [
    "homo sapiens", "primate", "bipedality",
    "universe comprises", "cosmology", "galactic",
    "english grammar", "adverb", "preposition",
    "dictionary is a listing", "lexemes"
]

blacklist = []
for u_doc in unique_docs:
    text_lower = u_doc["text"].lower()
    filename_lower = u_doc["filename"].lower()
    is_blacklist = False
    reason = ""

    for keyword in non_engineering_keywords:
        if keyword in text_lower:
            is_blacklist = True
            reason = f"Matches unrelated domain keyword: '{keyword}'"
            break

    if not is_blacklist:
        if len(u_doc["text"].strip()) < 200:
            is_blacklist = True
            reason = "Too short / empty placeholder"
        elif "gnu bash" in text_lower and "shell commands" in text_lower:
            is_blacklist = True
            reason = "Shell environment help output"
        elif "ubuntu desktop guide" in text_lower:
            is_blacklist = True
            reason = "OS general user manual"

    if is_blacklist:
        blacklist.append({
            "index": u_doc["index"],
            "id": u_doc["id"],
            "filename": u_doc["filename"],
            "reason": reason,
            "preview": u_doc["text"][:150].replace('\n', ' ') + "..."
        })

# Save the audit blacklist mapping
blacklist_path = base_dir / "document_integrity_blacklist.json"
blacklist_data = {
    "blacklist_indices": [item["index"] for item in blacklist],
    "blacklist_details": blacklist,
    "duplicate_details": duplicates
}
with blacklist_path.open("w", encoding="utf-8") as f:
    json.dump(blacklist_data, f, indent=2, ensure_ascii=False)

# 4. Generate Clean V2 Dataset and Decision Trail Manifest
cleaned_dir = base_dir / "cleaned_v2"
cleaned_dir.mkdir(parents=True, exist_ok=True)

blacklist_indices = set(blacklist_data["blacklist_indices"])
duplicates_info = blacklist_data["duplicate_details"]

duplicate_map = {item["index"]: item for item in duplicates_info}
blacklist_map = {item["index"]: item for item in blacklist_data["blacklist_details"]}

decision_trail = []
preserved_count = 0

for i, doc in enumerate(documents, start=1):
    doc_id = doc.get("id", f"index_{i}")
    filename = doc.get("fileName") or "NO_FILENAME"

    status = "keep"
    justification = "Preserved core engineering/relevant documentation."

    if i in blacklist_indices:
        status = "exclude"
        justification = blacklist_map.get(i, {}).get("reason", "Marked in the blacklist as unrelated material.")
    elif i in duplicate_map:
        status = "exclude"
        original_idx = duplicate_map[i]["duplicate_of_index"]
        justification = f"Exact content duplicate of Document #{original_idx}."

    decision_trail.append({
        "original_index": i,
        "document_id": doc_id,
        "filename": filename,
        "status": status,
        "justification": justification
    })

    if status == "keep":
        preserved_count += 1
        out_file = cleaned_dir / f"doc_{i:02d}_{filename.replace('/', '_')}.json"
        with out_file.open("w", encoding="utf-8") as out_f:
            json.dump(doc, out_f, ensure_ascii=False, indent=2)

decision_trail_path = base_dir / "decision_trail_manifest.json"
with decision_trail_path.open("w", encoding="utf-8") as f:
    json.dump(decision_trail, f, ensure_ascii=False, indent=2)

print("=== GENERATED CLEAN V2 DATASET ===")
print(f"Total processed documents: {len(documents)}")
print(f"Preserved documents written to cleaned_v2: {preserved_count}")
print(f"Excluded documents: {len(documents) - preserved_count}")
print(f"Blacklist saved: {blacklist_path}")
print(f"Decision trail manifest saved: {decision_trail_path}")
```

**Reasoning**:
I will load the verified QA benchmark and perform the side-by-side evaluation of the base model versus the adapted model in 4-bit precision, then save the comparison.



```python
import gc
import json
import torch
from pathlib import Path
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig
from peft import PeftModel

# 1. Define Paths and Variables
BASE_MODEL = "Qwen/Qwen2.5-1.5B-Instruct"
PROJECT_DIR = Path("/content/drive/MyDrive/SZM_AI_Training")
BENCHMARK_PATH = PROJECT_DIR / "verified_qa_benchmark.json"
ADAPTER_DIR = PROJECT_DIR / "szm-ai-lora-adapter"
RESULTS_PATH = PROJECT_DIR / "szm_benchmark_evaluation_results.json"

assert BENCHMARK_PATH.exists(), f"Benchmark file missing: {BENCHMARK_PATH}"
assert ADAPTER_DIR.exists(), f"Adapter directory missing: {ADAPTER_DIR}"

# 2. Load Benchmark
with BENCHMARK_PATH.open("r", encoding="utf-8") as f:
    benchmark = json.load(f)

# 3. Setup Quantization Config
quant_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True,
)

# 4. Define Generation Helper
def generate_answer(tok, model, question):
    messages = [
        {
            "role": "system",
            "content": (
                "You are SZM AI, an engineering assistant. "
                "Answer clearly. If uncertain, say so."
            ),
        },
        {"role": "user", "content": question},
    ]
    prompt = tok.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )
    inputs = tok(prompt, return_tensors="pt").to("cuda")

    with torch.inference_mode():
        output = model.generate(
            **inputs,
            max_new_tokens=150,
            do_sample=False,
            pad_token_id=tok.eos_token_id,
        )
    new_tokens = output[0, inputs["input_ids"].shape[1]:]
    return tok.decode(new_tokens, skip_special_tokens=True).strip()

# 5. Load Base Model and Run Benchmark
print("--- Loading Untouched Base Model ---")
tok = AutoTokenizer.from_pretrained(BASE_MODEL)
if tok.pad_token is None:
    tok.pad_token = tok.eos_token

base_model = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL,
    quantization_config=quant_config,
    device_map="auto",
)

base_results = []
for item in benchmark:
    print(f"Evaluating base model on: {item['question']}")
    ans = generate_answer(tok, base_model, item["question"])
    base_results.append(ans)

# 6. Free up GPU memory
del base_model
gc.collect()
torch.cuda.empty_cache()

# 7. Load Base Model + LoRA Adapter
print("--- Loading Base Model + Trained Adapter ---")
tok_adapted = AutoTokenizer.from_pretrained(ADAPTER_DIR)
if tok_adapted.pad_token is None:
    tok_adapted.pad_token = tok_adapted.eos_token

base_model_for_peft = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL,
    quantization_config=quant_config,
    device_map="auto",
)
adapted_model = PeftModel.from_pretrained(base_model_for_peft, str(ADAPTER_DIR))
adapted_model.eval()

adapted_results = []
for item in benchmark:
    print(f"Evaluating adapter on: {item['question']}")
    ans = generate_answer(tok_adapted, adapted_model, item["question"])
    adapted_results.append(ans)

# 8. Compile and Save Comparative Results
comparative_data = []
for i, item in enumerate(benchmark):
    comparative_data.append({
        "id": item["id"],
        "question": item["question"],
        "reference_answer": item["reference_answer"],
        "expected_citation": item["expected_citation"],
        "base_model_answer": base_results[i],
        "adapter_model_answer": adapted_results[i]
    })

with RESULTS_PATH.open("w", encoding="utf-8") as f:
    json.dump(comparative_data, f, indent=2, ensure_ascii=False)

print(f"Comparative results compiled and saved to: {RESULTS_PATH}")
for item in comparative_data:
    print(f"\n[Question]: {item['question']}")
    print(f"  Reference: {item['reference_answer']}")
    print(f"  Base Model: {item['base_model_answer']}")
    print(f"  Adapted Model: {item['adapter_model_answer']}")
```

```python
import json
from pathlib import Path

root = Path("/content/drive/MyDrive/SZM_AI_Training")
data = json.loads(
    (root / "szm-trained-local-corpus.json").read_text(encoding="utf-8")
)

terms = ["SZM Forge", "SZM Vitalis", "SZM Forge", "digital twin",
         "forging temperature", "Remaining Useful Life"]

print("===== ORIGINAL CORPUS SEARCH =====")
for term in terms:
    matches = []
    for i, doc in enumerate(data.get("documents", []), 1):
        text = doc.get("extractedText", "")
        if term.lower() in text.lower():
            matches.append((i, doc.get("fileName", "NO_FILENAME")))
    print(f"\n{term}: {len(matches)} matching documents")
    for match in matches[:10]:
        print(" ", match)

print("\n===== CLEANED V2 FILE CONTENTS =====")
clean = root / "cleaned_v2"

for p in sorted(clean.glob("*.json")):
    try:
        obj = json.loads(p.read_text(encoding="utf-8"))
        # Search the complete serialized object, regardless of its schema.
        text = json.dumps(obj, ensure_ascii=False)
        if any(term.lower() in text.lower()
               for term in ["SZM Forge", "SZM Vitalis"]):
            print(p.name, "contains project-name text")
    except Exception as e:
        print("Read error:", p.name, e)
```
